```python
import pandas as pd

# 1. Load the original small dataset
df = pd.read_csv('sample_pbft.csv')
print(f"Original shape: {df.shape}")

# 2. Duplicate the dataframe 1000 times
# ignore_index=True ensures the new index goes from 0 to 12,999 cleanly
df_large = pd.concat([df] * 10000, ignore_index=True)

# 3. OPTIONAL BUT HIGHLY RECOMMENDED: Shuffle the rows
# If you don't shuffle, your model will see the exact same 13 rows in the 
# exact same order 1000 times, which can cause cyclic learning patterns.
df_large = df_large.sample(frac=1.0, random_state=42).reset_index(drop=True)

print(f"New shape: {df_large.shape}")

# 4. Save to a new CSV
df_large.to_csv('sample_pbft_large.csv', index=False)
print("Saved to sample_pbft_large.csv")
```
